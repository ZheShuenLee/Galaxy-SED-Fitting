```python
# lets make a function for histogram comparison!!
import matplotlib.pyplot as plt
import numpy as np

def plothistogram(datasets, prop_name, bins=50, colors=None):
    """
    datasets: dict of {label: data_array}, e.g. {'base1': base1, 'base2': base2, 'base3': base3}
    """
    if colors is None:
        cmap = plt.get_cmap('tab10')
        colors = {label: cmap(i) for i, label in enumerate(datasets.keys())}
    
    # shared bin edges across ALL datasets
    all_mins = [np.nanmin(d) for d in datasets.values()]
    all_maxs = [np.nanmax(d) for d in datasets.values()]
    bin_edges = np.linspace(min(all_mins), max(all_maxs), bins + 1)
    
    plt.figure(figsize=(8, 6))
    
    for label, data in datasets.items():
        median = np.nanmedian(data)
        plt.hist(data, density=False, bins=bin_edges, histtype="step",
                  linewidth=1.5, label=label, color=colors[label])
        plt.axvline(median, color=colors[label], linestyle='--',
                     label=f'{label} Median = {median:.2e}')
    
    plt.ylabel('Counts')
    plt.xlabel(prop_name)
    plt.legend()
    plt.show()
```

```python
# let's do ks test!
from scipy import stats
def kstest(prop1, prop2):
    stat, p = stats.kstest(prop1, prop2, alternative='two-sided', nan_policy='omit')
    print('D-statistic=%.3f, p=%.5f' % (stat, p))
    if p > 0.05:
        print("Samples may come from the same distribution (fail to reject H_0)")
    else:
        print("Samples come from different distributions (reject H_0)")
```

```python
# plot ECDF for two sets of the same props
def ecdf(data):
    """Compute ECDF for a one-dimensional array of measurements."""
    # Number of data points: n
    n = len(data)
 
    # x-data for the ECDF: x
    x = np.sort(data)
 
    # y-data for the ECDF: y
    y = np.arange(1, n+1) / n
 
    return x, y

def plot_ecdf(datasets, prop_name, colors=None):
    """
    datasets: dict of {label: data_array}, e.g. {'base1': base1, 'base2': base2, 'base3': base3}
    """
    if colors is None:
        cmap = plt.get_cmap('tab10')
        colors = {label: cmap(i) for i, label in enumerate(datasets.keys())}
    
    plt.figure(figsize=(8, 6))
    
    for label, data in datasets.items():
        x, y = ecdf(data)
        plt.plot(x, y, color=colors[label], label=label)
    
    plt.ylabel('Probability')
    plt.xlabel(prop_name)
    plt.legend()
    plt.show()
```

```python
merge12 = pd.merge(base_1, base_2, left_on='id', right_on='id')
merge13 = pd.merge(base_1, base_3, left_on='id', right_on='id')
merge14 = pd.merge(base_1, base_4, left_on='id', right_on='id')
```

```python
# property name -> (base column name, error column name)
properties = {
    'Galaxy Mass (Msol)':  ('Galaxy Mass (solMass)',              'Galaxy Mass Error (solMass)'),
    'SFR (Msol yr-1)':           ('Star Formation Rate (solMass yr-1)', 'Star Formation Rate Error (solMass yr-1)'),
    'Dust Mass (Msol)':     ('Total Dust Mass (solMass)',          'Total Dust Mass Error (solMass)'),
    'Av (mag)':            ('Av (mag)',                           'Av Error (mag)'),
    'Galaxy Age (Myr)':    ('Average Galaxy Age (Myr)',           'Average Galaxy Age Error (Myr)'),
}

merges = {'merge12': merge12, 'merge13': merge13, 'merge14': merge14}
trim = {}

diff_cols = []
for prop_name in properties:
    diff_cols.append(f'{prop_name} difference')
    diff_cols.append(f'{prop_name} difference error')

for merge_name, merge in merges.items():
    for prop_name, (val_col, err_col) in properties.items():
        merge[f'{prop_name} difference'] = merge[f'{val_col}_x'] - merge[f'{val_col}_y']
        merge[f'{prop_name} difference error'] = np.sqrt(
            merge[f'{err_col}_x']**2 + merge[f'{err_col}_y']**2
        )
    trim[merge_name] = merge[diff_cols]

merge12 = trim['merge12']
merge13 = trim['merge13']
merge14 = trim['merge14']
```

```python
value_cols = [
    'Galaxy Mass (Msol) difference', 'SFR (Msol yr-1) difference',
    'Dust Mass (Msol) difference', 'Av (mag) difference', 'Galaxy Age (Myr) difference'
]
error_cols = [
    'Galaxy Mass (Msol) difference error', 'SFR (Msol yr-1) difference error',
    'Dust Mass (Msol) difference error', 'Av (mag) difference error', 'Galaxy Age (Myr) difference error'
]

merge12averaged = pd.DataFrame()

for col in value_cols:
    merge12averaged[col] = [merge12[col].mean()]

for col in error_cols:
    n = merge12[col].count()  # count, not len, to skip NaNs
    merge12averaged[col] = [np.sqrt((merge12[col]**2).sum()) / n]

merge12averaged.to_csv('merge12averaged.csv', index=False)
merge12averaged
```
